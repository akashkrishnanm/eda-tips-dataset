# Exploratory Data Analysis on Tips Dataset

## Import Libraries

```python
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt
```

## Load Dataset

```python
df = pd.read_csv("tips.csv")
```

## Data Overview

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

## Data Analysis

```python
df["tip"].mean()
```

```python
df["total_bill"].mean()
```

```python
df.groupby("day")["tip"].mean()
```

## Visualization

```python
sns.barplot(x="day", y="tip", data=df)
plt.show()
```

```python
sns.barplot(x="day", y="tip", data=df)
plt.title("Average Tip by Day")
plt.xlabel("Day")
plt.ylabel("Average Tip")
plt.show()
```

## Conclusion

- Sunday has the highest average tips  
- Tips tend to increase with total bill amount  
- Weekend days show higher tipping behavior compared to weekdays
